```python
import pandas as pd
import numpy as np
import os
import matplotlib.pyplot as plt
from sklearn.decomposition import PCA
from sklearn.decomposition import SparsePCA
from sklearn.preprocessing import StandardScaler
import seaborn as sns
from matplotlib.patches import Patch

sns.set_style('whitegrid')
sns.set_palette('colorblind')
palette = sns.color_palette('colorblind')
plt.rcParams.update({
    'font.size': 12,          # base font size
    'axes.titlesize': 26,     # subplot title
    'axes.labelsize': 22,     # axis label
    'xtick.labelsize': 20,    # tick label
    'ytick.labelsize': 20
})
network_colors = {
    'DMN': "#fb2e2e", 'VAN': '#40cce9', 'LANG': '#40cce9', 'TPOLE': '#2e87ae',
    'AUD': '#ce8fff', 'AMN': '#8a2edb', 'DAN': '#2efe2e',  'FP': '#ffff2e', 
    'MTL': '#89fd89', 'PMN': '#2e2eff', 'PON': "#ebebeb", 'SAL': '#2e2e2e',
    'SMd': '#8fffff', 'SMl': '#ffa12e', 'VIS': '#2e2eb3', 'MTL': '#89fd89'
    }
```

```python
df = pd.read_csv('/Users/emk/Documents/Documents - Ron Weasley V/Research/Verbal-Ability/Final/abcd_pMTG_FC_data_midb61_meanFC.csv')
```

```python
# Select Features
resid_cols = [col for col in df.columns if col.endswith('_resid') and 'full' not in col]
print(f"Found {len(resid_cols)} features for PCA.")

# Mean center the data (no normalization to unit variance)
df[resid_cols] = df[resid_cols] - df[resid_cols].mean()

# Extract matrix
X = df[resid_cols].values
n_features = X.shape[1]

# Reference PCA for sign consistency
pca_ref = PCA(random_state=42)
pca_ref.fit(X)
reference_components = pca_ref.components_
n_components = reference_components.shape[0]

# Convert component weights to loadings
true_loadings = pca_ref.components_.T * np.sqrt(pca_ref.explained_variance_)
loadings = true_loadings.T

# Bootstrapping
n_boot = 50000

boot_loadings = np.zeros((n_boot, n_components, n_features))
boot_var_exp = np.zeros((n_boot, n_components))

for b in range(n_boot):
    sample_idx = np.random.choice(X.shape[0], size=X.shape[0], replace=True)
    X_sample = X[sample_idx, :]

    pca = PCA(random_state=42)
    pca.fit(X_sample)
    boot_comp = pca.components_

    boot_var_exp[b, :] = pca.explained_variance_ratio_

    boot_true = boot_comp.T * np.sqrt(pca.explained_variance_)
    boot_true = boot_true.T

    for pc in range(n_components):
        if np.corrcoef(reference_components[pc], boot_comp[pc])[0, 1] < 0:
            boot_true[pc] *= -1

    boot_loadings[b, :, :] = boot_true

# 95% Confidence Intervals for single-feature loadings
ci_lower = np.percentile(boot_loadings, 2.5, axis=0)
ci_upper = np.percentile(boot_loadings, 97.5, axis=0)

# Full-data PCA outputs
var_exp = pca_ref.explained_variance_ratio_
var_exp_std = np.std(boot_var_exp, axis=0)

n_pcs = loadings.shape[0]

# Variance explained per PC
print("\nVariance explained per PC:")
for pc in range(min(10, n_pcs)):
    print(f"PC{pc+1}: {var_exp[pc]*100:.2f}%  (± {var_exp_std[pc]*100:.2f}%)")

# Results by hemispheric tag (_L_ / _R_) for printing
L_features = [i for i, f in enumerate(resid_cols) if "_L_" in f]
R_features = [i for i, f in enumerate(resid_cols) if "_R_" in f]

print(f"\nIdentified {len(L_features)} left pMTG")
print(f"Identified {len(R_features)} right pMTG")

print("\n===== OUTPUT SEPARATED BY _L_ AND _R_ FEATURES =====")

network_colors = {
    'DMN': "#fb2e2e", 'VAN': '#40cce9', 'LANG': '#40cce9', 'TPOLE': '#2e87ae',
    'AUD': '#ce8fff', 'AMN': '#8a2edb', 'DAN': '#2efe2e', 'FP': '#ffff2e',
    'MTL': '#89fd89', 'PMN': '#2e2eff', 'PON': "#ebebeb", 'SAL': '#2e2e2e',
    'SMd': '#8fffff', 'SMl': '#ffa12e', 'VIS': '#2e2eb3'
}

def parsed_parts(col):
    parts = col.split('_')
    return parts[0], parts[1]

def plot_label_for(base_net, net_hemi):
    b = base_net.upper()
    nh = net_hemi.lower()

    if b == "CO":
        b = "AMN"
    if b == "SMD":
        b = "SMd"
    if b == "SML":
        b = "SMl"
    if b == "TPOLE":
        b = "TPOLE"

    if b == "VAN" and nh == "left":
        return "LANG"
    if b == "VAN" and nh == "right":
        return "VAN"

    return b

def pick_index(base_net, net_hemi, pmtg_hemi):
    token = f"_{pmtg_hemi}_"
    matches = [
        i for i, c in enumerate(resid_cols)
        if parsed_parts(c)[0].upper() == base_net.upper()
        and parsed_parts(c)[1].lower() == net_hemi.lower()
        and token in c
        and c.endswith("_resid")
    ]
    return matches[0]

base_nets_present = sorted({parsed_parts(c)[0].upper() for c in resid_cols})

# Left/right network hemisphere plot groups
lr_plot_groups = []
for b in base_nets_present:
    if b == "VAN":
        lr_plot_groups.append((b, "left", "LANG"))
        lr_plot_groups.append((b, "right", "VAN"))
    else:
        lr_plot_groups.append((b, "left", plot_label_for(b, "left")))
        lr_plot_groups.append((b, "right", plot_label_for(b, "right")))

# Averaged network plot groups
mean_plot_groups = []
for b in base_nets_present:
    if b == "VAN":
        mean_plot_groups.append((b, "left_avg", "LANG"))
        mean_plot_groups.append((b, "right_avg", "VAN"))
    else:
        mean_plot_groups.append((b, "avg", plot_label_for(b, "left")))

# Tick labels for left/right plot
lr_xlabels = []
for _, hemi, lab in lr_plot_groups:
    if lab in ["LANG", "VAN"]:
        lr_xlabels.append(lab)
    else:
        lr_xlabels.append(f"{lab}\n{hemi}")

# Tick labels for mean plot
mean_xlabels = [lab for _, _, lab in mean_plot_groups]

lr_xtick_fontsize = 12

for pc in range(min(10, n_pcs)):
    print(f"\n=========== PC{pc+1} ===========\n")

    print("Top left pMTG features:")
    L_load = np.abs(loadings[pc, L_features])
    L_sorted = np.argsort(L_load)[-12:][::-1]
    for idx in L_sorted:
        fidx = L_features[idx]
        feature = resid_cols[fidx]
        mean_loading = loadings[pc, fidx]
        lower = ci_lower[pc, fidx]
        upper = ci_upper[pc, fidx]
        print(f"  {feature}: {mean_loading:.4f} [{lower:.4f}, {upper:.4f}]")

    print("\nTop right pMTG features:")
    R_load = np.abs(loadings[pc, R_features])
    R_sorted = np.argsort(R_load)[-12:][::-1]
    for idx in R_sorted:
        fidx = R_features[idx]
        feature = resid_cols[fidx]
        mean_loading = loadings[pc, fidx]
        lower = ci_lower[pc, fidx]
        upper = ci_upper[pc, fidx]
        print(f"  {feature}: {mean_loading:.4f} [{lower:.4f}, {upper:.4f}]")

    # Plot 1: left and right network hemispheres
    x_lr = np.arange(len(lr_plot_groups))
    w = 0.38

    pmtgL_means_lr = []
    pmtgR_means_lr = []
    pmtgL_err_lo_lr = []
    pmtgL_err_hi_lr = []
    pmtgR_err_lo_lr = []
    pmtgR_err_hi_lr = []

    for base_net, net_hemi, lab in lr_plot_groups:
        iL = pick_index(base_net, net_hemi, "L")
        iR = pick_index(base_net, net_hemi, "R")

        mL = loadings[pc, iL]
        mR = loadings[pc, iR]

        loL, hiL = np.percentile(boot_loadings[:, pc, iL], [2.5, 97.5])
        loR, hiR = np.percentile(boot_loadings[:, pc, iR], [2.5, 97.5])

        pmtgL_means_lr.append(mL)
        pmtgL_err_lo_lr.append(mL - loL)
        pmtgL_err_hi_lr.append(hiL - mL)

        pmtgR_means_lr.append(mR)
        pmtgR_err_lo_lr.append(mR - loR)
        pmtgR_err_hi_lr.append(hiR - mR)

    plt.figure(figsize=(18, 6))

    plt.bar(
        x_lr - w / 2,
        pmtgL_means_lr,
        w,
        yerr=[pmtgL_err_lo_lr, pmtgL_err_hi_lr],
        capsize=4,
        label='Left pMTG',
        color=[network_colors.get(lab, 'steelblue') for _, _, lab in lr_plot_groups],
        hatch='\\\\',
        edgecolor='black',
        linewidth=0.8
    )

    plt.bar(
        x_lr + w / 2,
        pmtgR_means_lr,
        w,
        yerr=[pmtgR_err_lo_lr, pmtgR_err_hi_lr],
        capsize=4,
        label='Right pMTG',
        color=[network_colors.get(lab, 'steelblue') for _, _, lab in lr_plot_groups],
        hatch='//',
        edgecolor='black',
        linewidth=0.8
    )

    plt.xticks(x_lr, lr_xlabels, rotation=45, ha='right', fontsize=lr_xtick_fontsize)
    plt.title(f'PC{pc + 1} Loadings')
    plt.ylabel('Correlation with PC')
    plt.xlabel('Network hemisphere')
    plt.axhline(0, color='black', linewidth=0.8)

    handles = [
        Patch(facecolor='white', edgecolor='black', hatch='\\\\', label='Left pMTG', linewidth=1.0),
        Patch(facecolor='white', edgecolor='black', hatch='//', label='Right pMTG', linewidth=1.0)
    ]
    leg = plt.legend(handles=handles, frameon=True)
    leg.get_frame().set_facecolor('white')
    leg.get_frame().set_edgecolor('black')

    plt.tight_layout()
    plt.show()

    # Plot 2: averaged across network hemispheres where appropriate
    x_mean = np.arange(len(mean_plot_groups))

    pmtgL_means_mean = []
    pmtgR_means_mean = []
    pmtgL_err_lo_mean = []
    pmtgL_err_hi_mean = []
    pmtgR_err_lo_mean = []
    pmtgR_err_hi_mean = []

    for base_net, mode, lab in mean_plot_groups:
        if mode == "avg":
            iL_left = pick_index(base_net, "left", "L")
            iL_right = pick_index(base_net, "right", "L")
            iR_left = pick_index(base_net, "left", "R")
            iR_right = pick_index(base_net, "right", "R")

            mL = 0.5 * (loadings[pc, iL_left] + loadings[pc, iL_right])
            mR = 0.5 * (loadings[pc, iR_left] + loadings[pc, iR_right])

            boot_avg_L = 0.5 * (
                boot_loadings[:, pc, iL_left] + boot_loadings[:, pc, iL_right]
            )
            boot_avg_R = 0.5 * (
                boot_loadings[:, pc, iR_left] + boot_loadings[:, pc, iR_right]
            )

            loL, hiL = np.percentile(boot_avg_L, [2.5, 97.5])
            loR, hiR = np.percentile(boot_avg_R, [2.5, 97.5])

        elif mode == "left_avg":
            iL = pick_index(base_net, "left", "L")
            iR = pick_index(base_net, "left", "R")

            mL = loadings[pc, iL]
            mR = loadings[pc, iR]

            loL, hiL = np.percentile(boot_loadings[:, pc, iL], [2.5, 97.5])
            loR, hiR = np.percentile(boot_loadings[:, pc, iR], [2.5, 97.5])

        elif mode == "right_avg":
            iL = pick_index(base_net, "right", "L")
            iR = pick_index(base_net, "right", "R")

            mL = loadings[pc, iL]
            mR = loadings[pc, iR]

            loL, hiL = np.percentile(boot_loadings[:, pc, iL], [2.5, 97.5])
            loR, hiR = np.percentile(boot_loadings[:, pc, iR], [2.5, 97.5])

        pmtgL_means_mean.append(mL)
        pmtgL_err_lo_mean.append(mL - loL)
        pmtgL_err_hi_mean.append(hiL - mL)

        pmtgR_means_mean.append(mR)
        pmtgR_err_lo_mean.append(mR - loR)
        pmtgR_err_hi_mean.append(hiR - mR)

    plt.figure(figsize=(16, 6))

    plt.bar(
        x_mean - w / 2,
        pmtgL_means_mean,
        w,
        yerr=[pmtgL_err_lo_mean, pmtgL_err_hi_mean],
        capsize=4,
        label='Left pMTG',
        color=[network_colors.get(lab, 'steelblue') for _, _, lab in mean_plot_groups],
        hatch='\\\\',
        edgecolor='black',
        linewidth=0.8
    )

    plt.bar(
        x_mean + w / 2,
        pmtgR_means_mean,
        w,
        yerr=[pmtgR_err_lo_mean, pmtgR_err_hi_mean],
        capsize=4,
        label='Right pMTG',
        color=[network_colors.get(lab, 'steelblue') for _, _, lab in mean_plot_groups],
        hatch='//',
        edgecolor='black',
        linewidth=0.8
    )

    plt.xticks(x_mean, mean_xlabels, rotation=45, ha='right')
    plt.title(f'PC{pc + 1} Loadings')
    plt.ylabel('Correlation with PC')
    plt.xlabel('Network')
    plt.axhline(0, color='black', linewidth=0.8)

    # handles = [
        # Patch(facecolor='white', edgecolor='black', hatch='\\\\', label='Left pMTG', linewidth=1.0),
        # Patch(facecolor='white', edgecolor='black', hatch='//', label='Right pMTG', linewidth=1.0)
    # ]
    # leg = plt.legend(handles=handles, frameon=True)
    # leg.get_frame().set_facecolor('white')
    # leg.get_frame().set_edgecolor('black')

    plt.tight_layout()
    plt.show()
```
